```python
cln_data = {}

names = [x.strip('#@') for x in data['Name']]

mapping = {
    "columbia university": "Columbia University",
    "new york university": "New York University",
    "city university of new york": "City University of New York",
    
}

alma_mater = [mapping[x.lower()] for x in data['Alma Mater']]
```

```python
data['Name'] = names
data['Alma Mater'] = alma_mater
```

```python
df = pd.DataFrame(data)
```

```python
alma_matter_ratios = df['Alma Mater'].value_counts()
plt.ylabel('Number of students')
plt.xlabel('Universities')
plt.title('Alma Mater')
plt.xticks(rotation=10)
plt.bar(alma_matter_ratios.index, alma_matter_ratios.values)
plt.show()
```
